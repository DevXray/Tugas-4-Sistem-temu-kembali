```python
# ─── Fungsi Preprocessing ────────────────────────────────────────────────
def preprocess_text(text):
   
    text = text.lower()

    text = re.sub(r'[^a-z\s]', ' ', text)  
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = text.split()

    tokens = [t for t in tokens if t not in stopword_set]

    tokens = [stemmer.stem(t) for t in tokens]

    return tokens

print('Fungsi preprocess_text() siap digunakan.')
```

```python
# Menerapkan ke Seluruh Dokumen
results = {}
for nama, teks in documents.items():
    token_sebelum = teks.split()              
    token_sesudah = preprocess_text(teks)     
    results[nama] = {
        'teks_asli'     : teks,
        'token_sebelum' : token_sebelum,
        'token_sesudah' : token_sesudah
    }

print('Preprocessing selesai untuk semua dokumen.')
```

```python
#Tabel Statistik Token
import pandas as pd

rows = []
for nama, data in results.items():
    n_sblm = len(data['token_sebelum'])
    n_sdh  = len(data['token_sesudah'])
    persen = round((n_sblm - n_sdh) / n_sblm * 100, 2)
    rows.append({
        'Dokumen'               : nama,
        'Token Sebelum'         : n_sblm,
        'Token Sesudah'         : n_sdh,
        'Pengurangan'           : n_sblm - n_sdh,
        'Pengurangan (%)' : persen
    })

df_stats = pd.DataFrame(rows)
print(df_stats.to_string(index=False))

# Rata-rata pengurangan
rata2 = df_stats['Pengurangan (%)'].mean()
print(f'\nRata-rata pengurangan token: {rata2:.2f}%')
```

## 6. Analisis Singkat

Preprocessing teks membawa dampak signifikan terhadap kualitas data dalam sistem Information Retrieval (IR). Berdasarkan hasil di atas, rata-rata sekitar 40–50% token dieliminasi melalui pipeline case folding, cleaning, stopwords removal, dan stemming. Eliminasi ini berdampak positif: noise berupa angka, tanda baca, dan kata-kata fungsional yang tidak membawa muatan semantik (seperti *dan*, *yang*, *di*, *untuk*) dihapus, sehingga vektor dokumen menjadi lebih ringkas dan representatif. Stemming selanjutnya menyatukan variasi morfologis seperti *pemeliharaan*, *dipelihara*, dan *terpelihara* ke akar kata *pelihara*, yang meningkatkan recall pada saat pencarian. Namun perlu dicatat bahwa over-stemming pada Sastrawi kadang menghasilkan stem yang tidak tepat untuk kata serapan (misalnya *malware* → *malwar*), sehingga untuk dokumen teknis berbahasa campuran, kualitas hasil stemming perlu diverifikasi secara manual.
